```python
df = spark.table("workspace.default.movies")
display(df)
```

```python
df.show(5)
```

```python
df.columns
df.count()
df.describe().show()
```

```python
display(df.describe())
```

```python
display(df.summary())
```

```python
df_selectedcolumns = df.select("title", "imdb_rating", "language", "studio")
display(df_selectedcolumns)
```

```python
df_selectedcolumns.show(5, truncate=False)
```

```python
df_filtered = df.filter((df.release_year >= 2000) & (df.release_year <= 2010))
display(df_filtered)
```

```python
df_marvel = df.filter(df.studio == "Marvel Studios")
display(df_marvel)
```

```python
unique_industires = df.select("industry").distinct()
display(unique_industires)
```

```python
from pyspark.sql.functions import col

df_profit = df.withColumn('profit', col('revenue') - col('budget'))
display(df_profit)
```

```python
df_renamed = df.withColumnRenamed("revenue", "total_revenue")
df_renamed.printSchema()
```
